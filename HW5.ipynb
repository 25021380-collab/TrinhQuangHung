**W5A16:** Viết hàm kiểm tra một số có phải số Armstrong hay không?

```python
n = int(input("Nhập số nguyên n: "))
s = str(abs(n))
p = len(s)
tong = sum(int(c) ** p for c in s)

if tong == abs(n):
    print("True")
else:
    print("False")
```

**W5A17:** Viết hàm kiểm tra một số có phải số chính phương hay không?

**W5A19:** Viết hàm kiểm tra một số có phải số kỳ diệu hay không.

```python
n = int(input("Nhập số nguyên n: "))
temp = abs(n)

while temp >= 10:
    tong = 0
    while temp > 0:
        tong += temp % 10
        temp //= 10
    temp = tong

if temp == 1:
    print("True")
else:
    print("False")
```

**W5A25:** Viết hàm point_in_quadrilateral(quad, p) kiểm tra điểm p có nằm trong hoặc trên biên của tứ giác không.

```python
print("Nhập tọa độ 4 đỉnh tứ giác (x y):")
quad = []
for i in range(4):
    pt = [float(x) for x in input(f"Đỉnh {i+1}: ").split()]
    quad.append(pt)

px = float(input("x của p: "))
py = float(input("y của p: "))

inside = False
n = 4
p1x, p1y = quad[0]
for i in range(n + 1):
    p2x, p2y = quad[i % n]
    if py > min(p1y, p2y):
        if py <= max(p1y, p2y):
            if px <= max(p1x, p2x):
                if p1y != p2y:
                    xinters = (py - p1y) * (p2x - p1x) / (p2y - p1y) + p1x
                if p1x == p2x or px <= xinters:
                    inside = not inside
    p1x, p1y = p2x, p2y

print(inside)
```

**W5A28:** Viết hàm tính góc ABC (độ) từ tọa độ 3 điểm A, B, C.

```python
import math

x1, y1 = float(input("xA: ")), float(input("yA: "))
x2, y2 = float(input("xB: ")), float(input("yB: "))
x3, y3 = float(input("xC: ")), float(input("yC: "))

ba_x, ba_y = x1 - x2, y1 - y2
bc_x, bc_y = x3 - x2, y3 - y2

dot = ba_x * bc_x + ba_y * bc_y
mag_ba = math.sqrt(ba_x**2 + ba_y**2)
mag_bc = math.sqrt(bc_x**2 + bc_y**2)

if mag_ba == 0 or mag_bc == 0:
    print("Không xác định")
else:
    cos_a = max(-1.0, min(1.0, dot / (mag_ba * mag_bc)))
    print(round(math.degrees(math.acos(cos_a)), 2))
```

**W5A33:** Viết hàm smart_taxi_fare tính tiền taxi.

```python
car_type = input("Loại xe (4_cho/7_cho): ").strip()
start_hour = int(input("Giờ chạy: "))
km = float(input("Số km: "))
wait_minutes = int(input("Phút chờ: "))

gia_mo_cua = 12000 if car_type == "4_cho" else 14000
gia_km = 15000 if car_type == "4_cho" else 17000
phi_cho = wait_minutes * 1000
tong = gia_mo_cua + (km * gia_km) + phi_cho

if start_hour >= 22 or start_hour < 5:
    tong *= 1.1
print(round(tong, 0))
```

**W5A34:** Viết hàm sum_1_to_n(n) tính tổng nếu các chữ số đôi một khác nhau.

```python
n = int(input("n = "))
if n <= 0:
    print(0)
else:
    tong = 0
    for x in range(1, n + 1):
        s = str(x)
        if len(s) == len(set(s)):
            tong += x
    print(tong)
```
