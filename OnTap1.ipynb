### Bài 1: Cước phí Taxi Thông Minh (smart_taxi_fare)

```python
car_type = input("Nhập loại xe (4_cho / 7_cho): ").strip()
start_hour = int(input("Nhập giờ khởi hành (0-23): "))
km = float(input("Nhập số km di chuyển: "))
wait_minutes = int(input("Nhập số phút chờ: "))

# 1. Tính tiền mở cửa và tiền di chuyển
if car_type == "4_cho":
    if km <= 0.8:
        phi_xe = 11000
    elif km <= 20:
        phi_xe = 11000 + (km - 0.8) * 13000
    else:
        phi_xe = 11000 + (20 - 0.8) * 13000 + (km - 20) * 10500
else:  # xe 7 cho
    if km <= 0.8:
        phi_xe = 12000
    elif km <= 20:
        phi_xe = 12000 + (km - 0.8) * 14000
    else:
        phi_xe = 12000 + (20 - 0.8) * 14000 + (km - 20) * 12000

# 2. Tính phí chờ
if wait_minutes <= 5:
    phi_cho = 0
else:
    import math
    phi_cho = math.ceil((wait_minutes - 5) / 5) * 2000

# Tổng tiền tạm tính
total = phi_xe + phi_cho
is_night = start_hour in [22, 23, 0, 1, 2, 3, 4, 5]

# 3. Áp dụng phụ thu ban đêm hoặc giảm giá đặc biệt
if is_night:
    total = total * 1.20
elif km > 30:
    total = total * 0.90

# 4. Điều kiện tối thiểu
if total < 50000 and km > 5:
    total = 50000

print("Tổng tiền:", int(round(total)))
```

### Bài 4: Tìm nghiệm thực phương trình bậc 2 (Sắp xếp tăng dần)

```python
import math
a = float(input("a = "))
b = float(input("b = "))
c = float(input("c = "))

if a == 0:
    if b != 0:
        print([-c / b])
    else:
        print([])
else:
    d = b*b - 4*a*c
    if d < 0:
        print([])
    elif d == 0:
        print([-b / (2*a)])
    else:
        r1 = (-b - math.sqrt(d)) / (2*a)
        r2 = (-b + math.sqrt(d)) / (2*a)
        print(sorted([r1, r2]))
```

### Bài 16: Tính góc ABC theo độ

```python
import math
x1, y1 = float(input("xA: ")), float(input("yA: "))
x2, y2 = float(input("xB: ")), float(input("yB: "))
x3, y3 = float(input("xC: ")), float(input("yC: "))

ba_x, ba_y = x1 - x2, y1 - y2
bc_x, bc_y = x3 - x2, y3 - y2

dot = ba_x * bc_x + ba_y * bc_y
mag_ba = math.sqrt(ba_x**2 + ba_y**2)
mag_bc = math.sqrt(bc_x**2 + bc_y**2)

cos_angle = max(-1.0, min(1.0, dot / (mag_ba * mag_bc)))
print(math.degrees(math.acos(cos_angle)))
```

### Bài 25: Kiểm tra điểm trong hoặc trên biên tứ giác (Ray-casting)

```python
print("Nhập tọa độ 4 đỉnh tứ giác (x y):")
quad = []
for i in range(4):
    quad.append([float(x) for x in input(f"Đỉnh {i+1}: ").split()])
px = float(input("x của p: "))
py = float(input("y của p: "))

inside = False
n = 4
p1x, p1y = quad[0]
for i in range(n + 1):
    p2x, p2y = quad[i % n]
    if py > min(p1y, p2y):
        if py <= max(p1y, p2y):
            if px <= max(p1x, p2x):
                if p1y != p2y:
                    xinters = (py - p1y) * (p2x - p1x) / (p2y - p1y) + p1x
                if p1x == p2x or px <= xinters:
                    inside = not inside
    p1x, p1y = p2x, p2y
print(inside)
```

### Bài 30: Nhận diện định dạng số (int, float, neither)

```python
s = input("Nhập chuỗi số s: ").strip()
if s.isdigit() or (s.startswith(('-', '+')) and s[1:].isdigit()):
    print("int")
else:
    if '.' in s:
        parts = s.split('.')
        if len(parts) == 2:
            p1 = parts[0].replace('-', '').replace('+', '')
            p2 = parts[1]
            if p1.isdigit() and p2.isdigit():
                print("float")
            else:
                print("neither")
        else:
            print("neither")
    else:
        print("neither")
```

### Bài 43: Tính diện tích tam giác theo công thức Heron
